O Conceito Técnico no Futebol
Para avaliar zagueiros e volantes, o analista não olha apenas o número bruto de desarmes. 
Precisamos entender a eficiência do desarme (quantos botes deram certo sem virar drible sofrido) e o volume de ações defensivas por partida.
Taxa de Sucesso em Desarmes (%):
Mede a precisão do bote defensivo:$$\text{taxa\_desarmes} = \left(\frac{\text{desarmes\_certos}}{\text{desarmes\_tentados}}\right) \times 100$$Total de Ações Defensivas:Soma das interceptações e dos desarmes bem-sucedidos:$$\text{acoes\_defensivas} = \text{desarmes\_certos} + \text{interceptacoes}$$Média de Ações por Jogo:Volume defensivo entregue por partida disputada:$$\text{acoes\_por\_jogo} = \frac{\text{acoes\_defensivas}}{\text{jogos}}$$

```python
ficha_defensor = {
    "Nome": "Luan Peres",
    "jogos": 38,
    "desarmes_tentados": 269,
    "desarmes_certos": 86,
    "interceptacoes": 92
}

taxa_desarmes = ficha_defensor["desarmes_certos"] / ficha_defensor["desarmes_tentados"] * 100
ficha_defensor["taxa_desarmes"] = taxa_desarmes

acoes_defensivas = ficha_defensor["desarmes_certos"] + ficha_defensor["interceptacoes"]
ficha_defensor["acoes_defensivas"] = acoes_defensivas

acoes_por_jogo = acoes_defensivas / ficha_defensor["jogos"]

print(f"Jogador: {ficha_defensor["Nome"]}, jogos disputados {ficha_defensor["jogos"]} ")
print(f"A taxa de sucesso nos desarmes foi de {taxa_desarmes:.2f} %")
print(f'A media de ações defensivas por jogo é {acoes_por_jogo:.2f} ')


```
