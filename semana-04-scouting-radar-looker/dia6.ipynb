```python
import pandas as pd
import matplotlib.pyplot as plt
from mplsoccer import PyPizza

parametros_lamine = ["Gols p90", "Assistências p90", "Chutes p90", 
    "Passes Chave p90", "Dribles p90", "Desarmes p90"
]

valores = [82, 91, 75, 88, 94, 45]

baker = PyPizza(
    params=parametros_lamine,
    straight_line_color="#000000",
    straight_line_lw=1,
    last_circle_lw=1,
    other_circle_lw=1,
    other_circle_ls="-."
)

# 4. Plotando a pizza
fig, ax = baker.make_pizza(
    valores,
    figsize=(8, 8),
    param_location=110,  # Distância dos nomes das fatias em relação ao centro
    kwargs_slices=dict(facecolor="royalblue", edgecolor="#000000", zorder=2, linewidth=1),
    kwargs_params=dict(color="#000000", fontsize=11, va="center"),
    kwargs_values=dict(color="#000000", fontsize=11, zorder=3,
    bbox=dict(edgecolor="#000000", facecolor="white", boxstyle="round,pad=0.2", lw=1))
)

# 5. Título tático
fig.text(0.515, 0.97, "Perfil de Scouting - Lamine Yamal", size=16, ha="center", weight="bold")
fig.text(0.515, 0.94, "Percentis calculados em comparação com liga", size=11, ha="center", color="gray")

plt.show()
```
