Entramos oficialmente na Semana 4 - Dia 5: Montagem de Dashboard Interativo de Scouting no Looker Studio com filtros de posição e idade.Hoje vamos transformar uma tabela simples em um produto dinâmico de inteligência de mercado. No futebol profissional, quando a diretoria define o perfil de contratação (ex: "precisamos de um jogador sub-23" ou "buscamos opções para o meio-campo com custo viável"), o analista não refaz o código do zero: ele entrega um painel onde a comissão filtra posições, idades e métricas em tempo real.A Conexão Tática: O Funil de RecrutamentoNo departamento de scouting, o recrutamento trabalha como um funil:Universo Amplo: Todos os atletas monitorados na base.Filtro de Perfil Tático: Seleção por posição (ex: apenas Extremos ou Volantes).Filtro de Política de Clube: Seleção por idade ou minutagem mínima (ex: atletas até 23 anos para revenda futura).Tomada de Decisão: Ordenação pelas métricas $p90$ de impacto (gols, assistências, dribles ou ações defensivas).Para alimentar essa inteligência, precisamos de uma base mais rica e diversificada gerada no Google Colab.

```python
import pandas as pd

# 1. Base ampliada de monitoramento de mercado
dados_mercado = {
    'jogador': [
        'Vinícius Júnior', 'Savinho', 'Jérémy Doku', 'Lamine Yamal',
        'Casemiro', 'Rodri', 'João Palhinha', 'Manuel Ugarte', 'André',
        'Jude Bellingham', 'Florian Wirtz', 'Cole Palmer', 'Gavi'
    ],
    'clube': [
        'Real Madrid', 'Manchester City', 'Manchester City', 'Barcelona',
        'Manchester United', 'Manchester City', 'Bayern de Munique', 'Manchester United', 'Wolverhampton',
        'Real Madrid', 'Bayer Leverkusen', 'Chelsea', 'Barcelona'
    ],
    'idade': [24, 20, 22, 17, 32, 28, 29, 23, 23, 21, 21, 22, 20],
    'posicao': [
        'Extremo', 'Extremo', 'Extremo', 'Extremo',
        'Volante', 'Volante', 'Volante', 'Volante', 'Volante',
        'Meia Ofensivo', 'Meia Ofensivo', 'Meia Ofensivo', 'Meia Ofensivo'
    ],
    'minutos_jogados': [1950, 1200, 980, 1420, 1420, 1890, 1100, 950, 1340, 1820, 1650, 1710, 890],
    'gols': [12, 5, 4, 6, 2, 6, 0, 0, 1, 15, 11, 14, 2],
    'assistencias': [7, 6, 3, 8, 2, 7, 1, 1, 1, 8, 12, 9, 3],
    'participacoes_gols': [19, 11, 7, 14, 4, 13, 1, 1, 2, 23, 23, 23, 5]
}

df_scouting = pd.DataFrame(dados_mercado)

# 2. Engenharia de Variáveis (Métricas p90)
df_scouting['gols_p90'] = ((df_scouting['gols'] / df_scouting['minutos_jogados']) * 90).round(2)
df_scouting['assistencias_p90'] = ((df_scouting['assistencias'] / df_scouting['minutos_jogados']) * 90).round(2)
df_scouting['part_gols_p90'] = ((df_scouting['participacoes_gols'] / df_scouting['minutos_jogados']) * 90).round(2)

# 3. Exportar base limpa
df_scouting.to_csv('base_scouting_interativo.csv', index=False, encoding='utf-8-sig')
print("Base gerada com sucesso! Linhas:", len(df_scouting))
```
