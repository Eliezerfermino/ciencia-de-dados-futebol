```python
def calcular_minutos_por_gol(minutos, gols):
    taxa = minutos / gols
    return taxa
```

```python
artilharia = [
    {"nome": "Gabriel", "minutos": 1600, "gols": 40},
    {"nome": "Neymar", "minutos": 987, "gols": 20},
    {"nome": "Rony", "minutos": 468, "gols": 10}
]

for atleta in artilharia:
    resultado = calcular_minutos_por_gol(atleta["minutos"],atleta["gols"])
    print(f"{atleta["nome"]} precisa de {resultado:.1f} para marcar um gol")
```
