```python
def calcular_minutos_por_gol(minutos, gols):
    resultado = minutos / gols
    return resultado

elenco_ataque = [
    {"nome": "Jogador A", "minutos": 900, "gols": 32},
    {"nome": "Jogador B", "minutos": 800, "gols": 28},
    {"nome": "Jogador C", "minutos": 846, "gols": 15},
    {"nome": "Jogador D", "minutos": 768, "gols": 11},
    {"nome": "Jogador E", "minutos": 498, "gols": 9},
    {"nome": "Jogador F", "minutos": 900, "gols": 7},
    {"nome": "Jogador G", "minutos": 800, "gols": 18},
    {"nome": "Jogador H", "minutos": 900, "gols": 21},
    {"nome": "Jogador I", "minutos": 359, "gols": 16},
    {"nome": "Jogador J", "minutos": 135, "gols": 7}
]
len(elenco_ataque)
print(f"Bem Vindo ao Santos FC!\nHoje vamos analisar o desempenho do nosso elenco de ataque.\nO time possui {len(elenco_ataque)} jogadores de ataque.\n")
for jogador in elenco_ataque:
    minutos_por_gol = calcular_minutos_por_gol(jogador["minutos"], jogador["gols"])
    if minutos_por_gol <= 100:
        status = "Desempenho de Elite"
    elif minutos_por_gol <= 200:
        status = "Desempenho Regular"
    else:
        status = "Desempenho Fraco"
        
    print(f"Jogador: {jogador['nome']}, Minutos por Gol: {minutos_por_gol:.1f}, Status: {status}")
    
    
```
