No futebol moderno, o camisa 10 ou o meia articulador precisa aliar volume a precisão. A comissão técnica monitora a taxa de acerto de passes para definir quem dita o ritmo da equipe e quem está entregando a posse de bola de graça para o adversário:Taxa de Acerto de Passes (%):$$\text{acerto} = \left(\frac{\text{passes\_certos}}{\text{passes\_tentados}}\right) \times 100$$Gatilhos de Decisão Tática com if / elif / else:Se a taxa for de 85% ou mais: o atleta é classificado como "Organizador de Elite" (alta segurança na posse).Se a taxa for de 75% até menos de 85%: o atleta recebe o status de "Distribuidor Regular" (bom rendimento, oscilação aceitável).Se a taxa for menor que 75%: acende o alerta de "Alerta: Alta Perda de Posse" (risco para a transição defensiva).Sua Missão no Google Colab


```python
meio_campo = [
    {"nome": "Mateus", "passes_tentados": 150, "passes_certos": 45 },
    {"nome": "Henrique", "passes_tentados": 200, "passes_certos": 156},
    {"nome": "João", "passes_tentados": 100, "passes_certos": 98 },
]

for atleta in meio_campo:
    precisao_passe = (atleta["passes_certos"] / atleta["passes_tentados"]) * 100
    if precisao_passe >= 85.0:
        status = "Organizador de Elite"
    elif precisao_passe > 75.0:
        status = 'Distribuidor Regular'
    else:
        status = "Alerta: Alta Perda de Posse"
        
    print(f"{atleta["nome"]}tem uma média de {precisao_passe:.1f} % na precisão do passe. \n Seu status é {status} ")
```
