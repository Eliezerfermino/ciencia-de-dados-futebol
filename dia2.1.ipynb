O Conceito Técnico no Futebol
No departamento de análise de mercado (scouting), raramente avaliamos um jogador isolado. O analista precisa comparar fichas técnicas de dois ou mais atletas para responder perguntas cruciais da comissão técnica:

Quem tem o melhor aproveitamento por minuto jogado?

Quem finaliza com maior eficácia em relação ao volume de jogo?

Para isso, trabalhamos com múltiplos dicionários e calculamos indicadores relativos para nivelar o tempo em campo.

```python
# Estruturando dois atletas com chaves padronizadas:


atleta_a = {"nome": "Jogador A", "minutos": 900, "gols": 6}
atleta_b = {"nome": "Jogador B", "minutos": 450, "gols": 4}
```

```python
# Calculando métricas relativas (ex: Minutos por Gol):

# Quanto menor o valor de minutos por gol, mais letal é o atleta:

min_por_gol_a = atleta_a["minutos"] / atleta_a["gols"]
min_por_gol_b = atleta_b["minutos"] / atleta_b["gols"]
```

```python
# Atualização de dados no Dicionário:

# Para salvar a nova métrica diretamente dentro da ficha de cada um:

atleta_a["min_por_gol"] = min_por_gol_a
atleta_b["min_por_gol"] = min_por_gol_b
```

```python
atacante_1 = {"Nome": "Neymar Junior", "Jogos": 50, "minutos": 4000, "chutes_gol": 335, "gols": 42}
atacante_2 = {"Nome": "Gabriel Barbosa", "Jogos": 42, "minutos": 3846, "chutes_gol": 286, "gols": 38}

taxa_conversao_1 = atacante_1["gols"] / atacante_1["chutes_gol"] * 100
taxa_conversao_2 = atacante_2["gols"] / atacante_2["chutes_gol"] * 100

atacante_1["taxa_conversao"] = taxa_conversao_1
atacante_2["taxa_conversao"] = taxa_conversao_2 

minutos_gol_1 = atacante_1["minutos"] / atacante_1["gols"]
minutos_gol_2 = atacante_2["minutos"] / atacante_2["gols"]

atacante_1["minutos_gol"] = minutos_gol_1
atacante_2["minutos_gol"] = minutos_gol_2

print(f'O jogador {atacante_1["Nome"]} obteve uma taxa de conversão de {atacante_1["taxa_conversao"]:.2f}% e {atacante_1["minutos_gol"]:.2f} para marcar')
print(f'O jogador {atacante_2["Nome"]} obteve uma taxa de conversão de {taxa_conversao_2:.2f}% e {minutos_gol_2:.2f} para marcar')
```
