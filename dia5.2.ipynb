No futebol de alto rendimento, o departamento de scout e análise tática monitora o aproveitamento em duelos de desarmes. Não basta apenas tentar dar o combate; o jogador precisa recuperar a posse de bola sem cometer faltas perigosas.Taxa de Eficiência em Desarmes (%):Mede o percentual de intervenções defensivas bem-sucedidas:$$\text{eficiencia} = \left(\frac{\text{desarmes\_certos}}{\text{desarmes\_tentados}}\right) \times 100$$Por que criar uma função (def):Em vez de reescrever essa fórmula toda vez que avaliar um defensor, você cria uma função chamada calcular_eficiencia_desarme(certos, tentados).Ela recebe os dois números, executa a divisão e multiplicação, e devolve com return o valor percentual pronto para o relatório ou para as condicionais de avaliação tática.

```python
def calcular_eficiencia_desarmes(certos, tentados):
    taxa = (certos / tentados) * 100
    return taxa

linha_defensiva = [
    {"nome": "Verissimo", "certos": 58, "tentados": 65},
    {"nome": "Luan", "certos": 35, "tentados": 65},
    {"nome": "Ananias", "certos": 15, "tentados": 65}
]

for atleta in linha_defensiva:
    eficiencia = calcular_eficiencia_desarmes(atleta["certos"], atleta["tentados"])
    if eficiencia >= 70.0:
        status = "Desarme de Elite"
    elif eficiencia >= 50.0:
        status = "Desarme Regular"
    else:
        status = "Alarme: Baixa Eficiencia Defensiva"
        
    print(f"{atleta["nome"]} - Eficiencia de Desarme: {eficiencia:.1f}% - Status: {status}")
    
```
