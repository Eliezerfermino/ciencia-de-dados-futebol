O Conceito Técnico no Futebol
No Dia 1, você aprendeu a criar listas com nomes de atletas. 
No Dia 2, estruturou dicionários com métricas individuais de cada jogador.

Agora, imagine que a diretoria do clube entregou a você a ficha estatística de todo o setor ofensivo. 
Calcular manualmente na calculadora a média de minutos por gol de cada atacante, um por um, consome muito tempo e gera erros.

É aqui que entra o laço for: ele funciona como o preparador físico passando de jogador em jogador na fila do teste ergométrico, aplicando o mesmo procedimento a todos.

```python
Lista de Dicionários:

É uma lista comum [], mas cada item dentro dela é um dicionário {} completo:

atacantes = [
    {"nome": "Neymar", "minutos": 900, "gols": 10},
    {"nome": "Gabigol", "minutos": 1200, "gols": 8},
    {"nome": "Marcos Leonardo", "minutos": 850, "gols": 5}
]
```

```python
O Laço for:

Ele percorre cada elemento da lista, dando a ele um nome temporário (por exemplo, atleta):

for atleta in atacantes:
    # Tudo que estiver com recuo (4 espaços) será executado para cada jogador
    minutos_por_gol = atleta["minutos"] / atleta["gols"]
    print(f"{atleta['nome']} precisa de {minutos_por_gol:.1f} minutos para marcar um gol.")
```

```python
eleno_ofensivo = [
    {"nome": "Gabriel Barbosa", "minutos": 650, "gols": 20},
    {"nome": "Neymar Jr", "minutos": 810, "gols": 12},
    {"nome": "Rony", "minutos": 723, "gols": 7}
]

for atleta in eleno_ofensivo:
    minutos_por_gols = atleta["minutos"] / atleta["gols"]
    print(f'{atleta["nome"]} fez {atleta["gols"]} gols. O jogador precisa de  {minutos_por_gols:.1f} minutos para marcar')
```
