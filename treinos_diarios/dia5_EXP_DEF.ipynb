Pense em uma função como uma jogada ensaiada no treino de bola parada. O treinador não explica a jogada inteira do zero a cada escanteio durante a partida; ele apenas chama o nome da jogada (por exemplo, "Jogada Escanteio 1") e os atletas executam os movimentos automaticamente.

Na programação é idêntico:

Em vez de reescrever a fórmula de porcentagem toda vez que avaliar um passe, você cria uma função chamada calcular_acerto_passe().

Sempre que você passar para ela os passes certos e tentados, ela devolve o resultado pronto.

```python
# Como declarar uma função (def):

# Usamos a palavra-chave def (de define), damos um nome claro para a função e definimos quais dados ela precisa receber entre parênteses (chamamos isso de parâmetros ou argumentos).

# O comando return é o passe final: ele devolve o resultado calculado para quem chamou a função.

def calcular_acerto_passe(passes_certos, passes_tentados):
    # Cálculo da porcentagem
    taxa = (passes_certos / passes_tentados) * 100
    return taxa
```

```python
# Como usar (chamar) a função:

# Você chama a função pelo nome, passa os valores reais e guarda o resultado em uma variável:

resultado = calcular_acerto_passe(85, 100)
print(f"Taxa de acerto: {resultado:.1f}%")
```

Vantagem no Futebol:

Se amanhã a diretoria pedir para mudar a fórmula ou arredondar de forma diferente, você altera apenas dentro da função, e todo o resto do seu sistema se atualiza sozinho.
