```python
def calcular_acerto_passe(passes_certos,passes_tentados):
    taxa = (passes_certos / passes_tentados) * 100
    return taxa


elenco = [
    {"nome": "Romeu", "passes_tentados": 58, "passes_certos": 32},
    {"nome": "Carlos", "passes_tentados": 79, "passes_certos": 49},
    {"nome": "Mateus", "passes_tentados": 31, "passes_certos": 28}
]

for atleta in elenco:
    resultado = calcular_acerto_passe (atleta["passes_certos"],atleta["passes_tentados"])
    if resultado >= 80:
        status = "Atleta em Destaque"
    elif resultado >= 50:
        status = "Atleta Regular"
    else:
        status = "Alerta: Atleta precisa de atenção"
        
    print(f"{atleta["nome"]} tem uma taxa de {resultado:.1f} acerto de passes. \n O status desse jogador é {status}")
```
