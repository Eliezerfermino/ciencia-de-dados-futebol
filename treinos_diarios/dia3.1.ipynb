O Conceito Técnico no FutebolNo Dia 3, você viu como percorrer uma lista de atacantes para calcular minutos por gol. No meio-campo, a comissão técnica quer avaliar dois pilares fundamentais:Segurança de Posse: a precisão nos passes.Capacidade Criativa: o volume de passes decisivos (assistências para finalização).Para automatizar essa análise sem precisar calcular jogador por jogador manualmente, usamos o laço for para percorrer cada atleta e calcular as taxas em lote:Taxa de Acerto de Passe (%):$$\text{acerto} = \left(\frac{\text{passes\_certos}}{\text{passes\_tentados}}\right) \times 100$$Passes Decisivos por Jogo:$$\text{decisivos\_por\_jogo} = \frac{\text{passes\_decisivos}}{\text{jogos}}$$Automação com for:O Python passa por cada meia, faz os cálculos usando as chaves do dicionário e exibe o boletim completo formatado.

```python
meio_campistas = [
    {"nome": "Rollheiser", "jogos": 25, "passes_tentados": 150, "passes_certos": 115, "passes_decisivos": 95},
    {"nome": "Schimidt", "jogos": 21, "passes_tentados": 168, "passes_certos": 101, "passes_decisivos": 84},
    {"nome": "Arthur", "jogos": 24, "passes_tentados": 186, "passes_certos": 154, "passes_decisivos": 124}
]

for jogadores in meio_campistas:
    acertos_passes = jogadores["passes_certos"] / jogadores["passes_tentados"] * 100
    decisivos_por_jogos = jogadores["passes_decisivos"] / jogadores["jogos"]
    print(f'{jogadores["nome"]} tem {acertos_passes:.1f} % nos passes certos com a media de {decisivos_por_jogos:.1f} passes decisivos por jogo')
   
```
