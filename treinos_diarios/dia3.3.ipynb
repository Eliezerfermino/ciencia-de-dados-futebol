O Conceito Técnico no FutebolNa análise de desempenho de goleiros, a comissão técnica não avalia apenas os gols sofridos. Um goleiro exposto a uma enxurrada de chutes pode tomar mais gols, mas ainda assim ter uma atuação brilhante. Por isso, duas métricas são fundamentais:Taxa de Defesas (%):Mede a eficácia contra finalizações no alvo:$$\text{taxa\_defesas} = \left(\frac{\text{defesas}}{\text{chutes\_recebidos}}\right) \times 100$$Média de Defesas por Jogo:Volume de intervenções exigidas por partida:$$\text{defesas\_por\_jogo} = \frac{\text{defesas}}{\text{jogos}}$$

```python
goleiros = [
    {"nome": "Brasão", "jogos": 45, "chutes_recebidos": 247, "defesas":215},
    {"nome": "João Paulo", "jogos": 45, "chutes_recebidos": 349, "defesas":298},
    {"nome": "Fabio", "jogos": 45, "chutes_recebidos": 186, "defesas":138}
]

for goleiro in goleiros:
    taxa_defesa = goleiro["defesas"] / goleiro["chutes_recebidos"] * 100
    media_defesas_jogo = goleiro["defesas"] / goleiro["jogos"]
    
    print(f'{goleiro["nome"]} tem {taxa_defesa:.1f} % em defesas com media de {media_defesas_jogo:.1f} defesas por jogos')
```
