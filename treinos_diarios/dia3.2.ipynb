O Conceito Técnico no FutebolNa análise tática de zagueiros e volantes, o departamento de desempenho precisa avaliar a consistência nas disputas de bola. Calcular jogador por jogador na mão demora muito. Usamos o laço for para processar a linha de defesa inteira em um instante.As métricas que a comissão técnica quer monitorar:Taxa de Eficiência nos Desarmes (%):$$\text{eficiencia} = \left(\frac{\text{desarmes\_certos}}{\text{desarmes\_tentados}}\right) \times 100$$Total de Ações Defensivas por Jogo:$$\text{acoes\_por\_jogo} = \frac{\text{desarmes\_certos} + \text{interceptacoes}}{\text{jogos}}$$

```python
linha_defensiva = [
    {"nome": "Luan Peres", "jogos": 28, "desarmes_tentados": 168, "desarmes_certos": 101, "interceptacoes": 58},
    {"nome": "Lucas Verissimo", "jogos": 27, "desarmes_tentados": 123, "desarmes_certos": 84, "interceptacoes": 35},
    {"nome": "Ananias", "jogos": 28, "desarmes_tentados": 101, "desarmes_certos": 74, "interceptacoes": 29}
]

for defensor in linha_defensiva:
    eficiencia_desarmes = defensor["desarmes_certos"] / defensor["desarmes_tentados"] * 100
    acoes_por_jogo = (defensor["desarmes_certos"] + defensor["interceptacoes"]) / defensor["jogos"]
    
    print(f'{defensor["nome"]} tem uma eficiencia de {eficiencia_desarmes:.1f} % em desarmes e uma media defensiva por jogo de {acoes_por_jogo:.1f}')
```
