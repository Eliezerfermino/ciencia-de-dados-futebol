```python
# Chave e Valor (chave: valor):

# O dicionário funciona como a ficha do departamento de análise: você define um rótulo (chave) e associa um dado a ele (valor).

# As chaves ficam entre aspas e são separadas do valor por dois pontos :. Cada par é separado por vírgula ,.


jogador = {
    "nome": "Neymar",
    "clube": "Santos",
    "gols": 14,
    "assistencias": 8,
    "minutos_jogados": 1250
}
# Acessando valores pelas chaves:

# Em vez de usar posições numéricas ([0], [1]), você busca a informação direto pelo nome do rótulo:


total_gols = jogador["gols"]  # Retorna 14
# Criando novas métricas calculadas:

# Podemos adicionar novas métricas à ficha do atleta fazendo cálculos diretos:


jogador["participacoes_diretas"] = jogador["gols"] + jogador["assistencias"]
```

```python
ficha_atleta = {
    "nome": "Gabriel Brasão",
    "posicao": "Goleiro",
    "minutos_jogados": 85,
    "gols": 15,
    "assistencias": 3
}

ficha_atleta["participacoes_em_gol"] = ficha_atleta["gols"] + ficha_atleta["assistencias"]

minutos_por_participação = ficha_atleta["minutos_jogados"] / ficha_atleta["participacoes_em_gol"]

print(f'O jogador em destaque é {ficha_atleta["nome"]} na posição {ficha_atleta["posicao"]}')
print("A média de participações em gols é ", ficha_atleta["participacoes_em_gol"])
print(f'A média de minutos necessarios para ele participar de um gol é {minutos_por_participação:.2f}')
ficha_atleta
```
