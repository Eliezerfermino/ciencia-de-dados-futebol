No departamento de inteligência e análise de desempenho de um clube, os analistas trabalham com gatilhos de alerta e relatórios de recomendação:

Se a média de desarmes certos por jogo de um volante for alta, ele recebe a recomendação de "Prioridade para Titularidade".

Se o número for intermediário, ele fica classificado como "Opção de Rotação / Banco".

Se estiver abaixo do esperado pela comissão técnica, acende o sinal de "Alerta: Necessita Ajuste de Marcação".


```python
volantes = [
    {"nome": "Augusto", "jogos": 50, "desarmes": 100 },
    {"nome": "Marcos", "jogos": 35, "desarmes": 150 },
    {"nome": "Samuel", "jogos": 15, "desarmes": 6 },
]

for atleta in volantes:
    desarmes_por_jogo = atleta["desarmes"] / atleta["jogos"]
    if desarmes_por_jogo >= 3.0:
        status = "Titular Absoluto"
    elif desarmes_por_jogo >= 1.5:
        status = "Opção de Elenco"
    else:
        status = "Alerta: Desempenho Abaixo da Média"
        
    print(f'{atleta["nome"]} tem uma média de {desarmes_por_jogo:.1f} desarmes por jogo. \n Seu status é {status}')
```
