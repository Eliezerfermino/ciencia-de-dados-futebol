No futebol profissional, a comissão técnica e o departamento de análise trabalham com gatilhos de decisão e alertas de desempenho:

Se um atacante precisa de menos de 100 minutos para marcar, ele recebe o status de "Artilheiro de Elite".

Se precisa de até 200 minutos, tem um desempenho "Regular / Produtivo".

Caso precise de mais tempo que isso, acende o alerta de "Baixa Efetividade / Atenção".

```python
minutos_por_gol = 85.5

if minutos_por_gol < 100:
    status = "Artilheiro de Elite"
elif minutos_por_gol <= 200:
    status = "Produtivo"
else:
    status = "Atenção: Baixa Efetividade"
```

```python
elenco_ofensivo = [
     {"nome": "Vini Jr", "minutos": 900, "gols": 10},
     {"nome": "Rodrygo", "minutos": 843, "gols": 6},
     {"nome": "Endrick", "minutos": 659, "gols": 2}
]

for atleta in elenco_ofensivo:
    minutos_gol = atleta["minutos"] / atleta["gols"]
    if minutos_gol < 90:
        alerta = "Desempenho de Elite"
    elif minutos_gol <= 180:
        alerta = "Desempenho Regular"
    else:
        alerta = "Alerta: Necessita Ajuste Técnico"
        
    print(f"{atleta["nome"]} tem uma média de {minutos_gol:.1f} minutos por gol. \nO desempenho dele é: {alerta}\n" ) 
    
```
