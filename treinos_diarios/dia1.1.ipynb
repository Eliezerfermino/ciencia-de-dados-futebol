### O Conceito Técnico no Futebol
No futebol, o elenco muda dinamicamente: um jogador pode ser contratado, vendido ou suspenso. 
Além disso, o analista precisa acessar não só o primeiro nome da lista, 
mas também o último (como o centroavante de referência) ou alterar quem veste determinada camisa.

```python
# Acessando o último jogador com índice negativo ([-1]):

#  Python permite contar de trás para frente. O índice [0] pega o primeiro, e o índice [-1] 
# acessa diretamente o último da lista sem você precisar saber o tamanho total dela.


atacantes = ["Vinicius Jr", "Rodrygo", "Mbappé"]
ultimo = atacantes[-1]  # Retorna "Mbappé"
```

```python
# Adicionando reforços com .append():

# Para incluir um novo atleta no fim da lista (uma nova contratação), usamos o método .append():

atacantes.append("Endrick")
```

```python
# Substituição direta pelo índice:

# Se um jogador sai lesionado ou é negociado, podemos trocar o valor daquela posição específica:

atacantes[0] = "Savinho"  # Substitui o elemento do índice 0
```

```python
titulares = ["G.Barbosa", "Alvaro Barreal", "João Paulo", "Escobar"]
print(f'{titulares[-1]}')
titulares.append("Phelipe Coutinho") # ADICIONA 
titulares[1] = 'Phelipe Coutinho'
titulares[-1] = "Miguelito" # ADICIONOU A LISTA SEM O .APPEND
tamanho_elenco = len(titulares)

print(f'{tamanho_elenco} são os relacionados para partida de hoje: {titulares}')
```
