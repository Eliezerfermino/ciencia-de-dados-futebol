## 🛠️ Metodologia e Ferramentas Técnicas
- **Ambiente:** Google Colab (processamento em nuvem com Python).
- **Manipulação de Dados:** `pandas`
- **Operações e Conceitos Aplicados:**
  - **Cálculo Vetorizado:** Criação das colunas `participacoes_diretas` (gols + assistências) e da taxa relativa `media_participacoes` por jogo, com arredondamento padronizado (`round(2)`).
  - **Agregações Multivariadas (`groupby` + `agg`):** Agrupamento por time combinando cálculos distintos em uma única passada (soma para gols e média para desarmes).
  - **Filtros Condicionais Múltiplos (`&`):** Isolamento de amostras com significância estatística (`jogos > 10` e `gols > 5`).
  - **Higienização de Estrutura:** Ordenação decrescente de rankings (`sort_values`) e reindexação limpa (`reset_index(drop=True)`).
  - **Apresentação Executiva:** Estruturação visual com `IPython.display.Markdown` para renderização de tabelas e seções limpas.

## 📊 Relatórios Gerados
- **Ranking Coletivo das Equipes:** Classificação dos clubes por volume de gols e consistência de desarme.
- **Painel de Destaques Individuais:** Lista restrita dos atletas mais decisivos do torneio aptos para relatórios de mercado.

O Relatório Executivo do Campeonato
A diretoria do torneio contratou você como analista de dados oficial para processar a base de atletas e entregar dois relatórios táticos:

O Ranking Coletivo das Equipes: mostrando o total de gols marcados e a média de desarmes de cada time.

O Painel de Destaques Individuais (Scouting de Ouro): identificando os jogadores mais eficientes e participativos do campeonato.

```python
import pandas as pd
from IPython.display import display, Markdown
dados_campeonato = {
    "atleta": [
        "Gabriel Barbosa", "Arrascaeta", "Pedro", 
        "Raphael Veiga", "Rony", "Gustavo Gómez", 
        "Yuri Alberto", "Renato Augusto", "Gil"
    ],
    "time": [
        "Flamengo", "Flamengo", "Flamengo", 
        "Palmeiras", "Palmeiras", "Palmeiras", 
        "Corinthians", "Corinthians", "Corinthians"
    ],
    "jogos": [12, 10, 11, 12, 11, 12, 10, 8, 12],
    "gols": [8, 4, 7, 6, 3, 2, 5, 2, 1],
    "assistencias": [2, 7, 1, 4, 3, 0, 1, 5, 0],
    "desarmes": [5, 12, 4, 18, 14, 32, 6, 11, 28]
}

df_campeonato = pd.DataFrame(dados_campeonato)

df_campeonato["participacoes_diretas"] = df_campeonato["assistencias"] + df_campeonato['gols']
df_campeonato['media_participacoes'] = (df_campeonato['participacoes_diretas'] / df_campeonato['jogos']).round(2) * 100

ranking_equipes = df_campeonato.groupby("time").agg({'gols': 'sum', 'desarmes': 'mean'}).round(2).reset_index()
ranking_equipes = ranking_equipes.sort_values(by='gols', ascending=False).reset_index(drop=True)

destaques_individuais = df_campeonato[(df_campeonato['jogos'] > 10) & (df_campeonato['gols'] > 5)]
destaques_individuais = destaques_individuais.sort_values(by='media_participacoes', ascending=False)

display(
    Markdown('###  Ranking Coletivo da Equipe'), 
    ranking_equipes,
    Markdown('### Destaques Individuais (Scoulting)'),
    destaques_individuais
)
```
