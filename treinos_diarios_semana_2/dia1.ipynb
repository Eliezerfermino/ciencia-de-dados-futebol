```python
# O Conceito de Hoje: Criação de DataFrames Manuais (Dia 1)
# Para usar o Pandas no Google Colab, primeiro importamos a biblioteca dando a ela o apelido padrão de mercado: pd.

import pandas as pd


# Podemos transformar um dicionário onde cada chave é o nome de uma coluna e cada valor é uma lista de dados diretamente em uma tabela usando pd.DataFrame():

dados = {
    "jogador": ["Neymar Jr", "Barreal", "Gabriel Barbosa"],
    "posicao": ["Ponta Esquerda", "Meia Armador", "Centroavante"],
    "gols": [138, 36, 84]
}

tabela_santos = pd.DataFrame(dados)
display(tabela_santos)
```
