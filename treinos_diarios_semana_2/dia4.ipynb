No futebol profissional, os números brutos muitas vezes enganam. Um jogador com 10 gols em 30 partidas tem uma eficiência diferente de um atacante que marcou 8 gols em apenas 10 jogos. Por isso, analistas de desempenho e scouts criam métricas derivadas (taxas e proporções) para equalizar a comparação entre os atletas.

No Pandas, criar uma coluna calculada é tão simples quanto fazer uma operação matemática comum, pois ele aplica a conta linha por linha automaticamente (uma operação vetorizada).

```python
# Para criar uma nova coluna em um DataFrame existente:

# Declaramos o nome da nova coluna entre colchetes: df["nova_coluna"].

# Atribuímos a ela o cálculo feito com colunas já existentes.

# Exemplo Prático: Média de Gols por Partida
# Se tivermos colunas de "gols" e "jogos", podemos calcular a taxa de gols por partida dividindo 
# uma coluna pela outra:

import pandas as pd

# Criando a nova coluna de média diretamente:
df["gols_por_jogo"] = df["gols"] / df["jogos"]

#Também podemos somar colunas. Por exemplo, a métrica clássica de participações diretas em gols:

df["participacoes_em_gols"] = df["gols"] + df["assistencias"]
```

```python
dados_atletas = {
    "Jogador": ["Jogador A", "Jogador B", "Jogador C", "Jogador D", "Jogador E"],
    "Jogos": [20, 25, 30, 15, 10],
    "Gols": [10, 15, 20, 5, 2],
    "Assistencias": [5, 7, 10, 3, 1]
}

df_ataque = pd.DataFrame(dados_atletas)
df_ataque["participacoes_diretas"] = df_ataque["Gols"] + df_ataque["Assistencias"]
df_ataque["media_participacoes"] = (df_ataque["participacoes_diretas"] / df_ataque["Jogos"]).round(2)

df_ataque = df_ataque.sort_values(by="media_participacoes", ascending=False)

display(df_ataque)
```
