```python
import pandas as pd

dados_passes = {
    "Jogador": ["Lucas", "Rodinei", "Coutinho", "Gabigol", "Rony"],
    "Posicao": ["Zagueiro", "Lateral", "Meia", "Atacante", "Atacante"],
    "Passes_Completos": [50, 70, 90, 80, 60],
    "Perdas_de_posse": [5, 10, 15, 8, 12]
}

df_construcao = pd.DataFrame(dados_passes)

df_distribuicao = df_construcao[["Jogador", "Posicao", "Passes_Completos"]]

ranking_distribuicao = df_distribuicao.sort_values(by="Passes_Completos", ascending=False)

display(ranking_distribuicao)
```
