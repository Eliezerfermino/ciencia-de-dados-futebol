Dia 3: Filtros condicionais múltiplos (operadores & e |)! ⚽🔍

No departamento de análise de mercado (scouting), dificilmente você buscará um jogador olhando apenas uma métrica isolada. Por exemplo: não basta o atacante ter muitos gols se ele jogou apenas 2 partidas, ou se ele atua em uma posição que não interessa à comissão técnica.

Precisamos combinar critérios! É aqui que entram os filtros condicionais no Pandas.

O Conceito Técnico
Para filtrar linhas em um DataFrame com base em condições, usamos operadores lógicos. No Pandas, existem duas regras essenciais que você não pode esquecer:

Operadores bitwise específicos:

Usamos o símbolo & para representar o E (ambas as condições precisam ser verdadeiras).

Usamos o símbolo | (pipe) para representar o OU (pelo menos uma das condições precisa ser verdadeira).
(Atenção: não usamos as palavras and e or no Pandas como fazíamos no Python puro).

Parênteses obrigatórios:

Cada condição individual deve estar envolvida por parênteses ( ). Sem eles, o Python se perde na ordem das operações e acusa erro.


```python
# Filtro combinado com E (&)
filtro_meias_criativos = df[(df["posicao"] == "Meia") & (df["assistencias"] > 3)]
display(filtro_meias_criativos)
```

```python
# Filtro combinado com OU (|)
filtro_defensores = df[(df["posicao"] == "Zagueiro") | (df["posicao"] == "Lateral")]
display(filtro_defensores)
```

```python
import pandas as pd

df_scoulting = {
    "Atleta": ["João", "Maria", "Pedro", "Ana", "Lucas","Antonio"],
    "Posição": ["Atacante", "Meia", "Defensor", "Goleiro", "Atacante","Meia"],
    "Gols": [10, 5, 2, 0, 8, 3],
    "Jogos": [20, 15, 18, 22, 25, 17]
}

df_scoulting = pd.DataFrame(df_scoulting)

# A diretoria solicitou um relatório urgente:
# nela quer mapear atletas que apresentem amostra confiável de minutos e alto poder de decisão.

destaques_ofensivos = df_scoulting[(df_scoulting["Jogos"] > 10) & (df_scoulting["Gols"] > 5)]

display(destaques_ofensivos)

```
