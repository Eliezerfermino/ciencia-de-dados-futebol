```python
import pandas as pd

dados_jogadores = {
    "Atleta": ["Cristiano Ronaldo", "Lionel Messi", "Neymar Jr", "Kylian Mbappé", "Kevin De Bruyne"],
    "chutes_totais": [150, 120, 100, 90, 80],
    "chutes_no_alvo": [80, 70, 60, 50, 40],
    "gols": [30, 25, 20, 15, 10],
    
}

df_finalizacoes = pd.DataFrame(dados_jogadores)

df_finalizacoes["precisao_chute"] = (df_finalizacoes["chutes_no_alvo"] / df_finalizacoes["chutes_totais"]).round(2) * 100
df_finalizacoes["taxa_conversao"] = (df_finalizacoes["gols"] / df_finalizacoes["chutes_no_alvo"]).round(2)

df_finalizacoes = df_finalizacoes.sort_values(by="precisao_chute", ascending=False)
display(df_finalizacoes)

```
