Vamos para mais um desafio prático de consolidação da Semana 2 - Dia 5: Agrupamentos e métricas por equipe (groupby, agg)! ⚽📊

No departamento de inteligência de um torneio, os analistas precisam frequentemente cruzar métricas defensivas e disciplinares por posição ou por time para entender o perfil de jogo de cada elenco.

O Cenário Tático
A coordenação técnica quer avaliar o comportamento disciplinar e a capacidade de desarme de três clubes rivais ao longo da competição. Você recebeu uma tabela com o registro de 6 atletas.

```python
import pandas as pd

dados_disciplina = {
    "atleta": ["Felipe Melo", "Gustavo Gómez", "Fernandinho", "Kannemann", "Arrascaeta", "Raphael Veiga"],
    "time": ["Fluminense", "Palmeiras", "Athletico-PR", "Grêmio", "Flamengo", "Palmeiras"],
    "cartoes_amarelos": [6, 4, 5, 7, 1, 2],
    "desarmes": [28, 35, 30, 42, 12, 18]
}

df_disciplina = pd.DataFrame(dados_disciplina)

resumo_disciplina = df_disciplina.groupby("time").agg({"cartoes_amarelos": "sum", "desarmes": "mean"}).round(2).reset_index()
resumo_disciplina = resumo_disciplina.sort_values(by="cartoes_amarelos", ascending=False).reset_index(drop=True)

display(resumo_disciplina)
```
