```python
import pandas as pd

dados_scouting = {
    "Atleta": ["Rony", "Gabriel Barbosa", "Raphael Veiga", "Hulk", "Pedro"],
    "Jogos": [28, 22, 30, 26, 25],
    "Minutos_Jogados": [2240, 1760, 2550, 2340, 2000],
    "Gols": [9, 12, 11, 15, 18],
    "Assistencias": [5, 3, 8, 6, 2]
}

df_scouting = pd.DataFrame(dados_scouting)

df_scouting["participacoes_diretas"] = df_scouting["Gols"] + df_scouting["Assistencias"]
df_scouting["minutos_por_participacao"] = (df_scouting["Minutos_Jogados"] / df_scouting["participacoes_diretas"]).round(2)

df_scouting = df_scouting.sort_values(by="minutos_por_participacao", ascending=True)

display(df_scouting)

```
